```python
import pandas as pd
from sklearn.model_selection import train_test_split

# 1. Charger les données
chemin = "data/raw/creditcard.csv"
df = pd.read_csv(chemin)

print("Dimensions avant suppression des doublons :", df.shape)

# 2. Supprimer les doublons exacts
df_propre = df.drop_duplicates().reset_index(drop=True)

print("Dimensions après suppression des doublons :", df_propre.shape)
print("Nombre de doublons supprimés :", len(df) - len(df_propre))

# 3. Séparer les variables explicatives et la cible
X = df_propre.drop("Class", axis=1)
y = df_propre["Class"]

print("Dimensions de X :", X.shape)
print("Dimensions de y :", y.shape)
print("Valeurs possibles de y :")
print(y.value_counts())

# 4. Séparer les données en entraînement et en test
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Dimensions de X_train :", X_train.shape)
print("Dimensions de X_test :", X_test.shape)
print("Nombre de fraudes dans y_train :", y_train.sum())
print("Nombre de fraudes dans y_test :", y_test.sum())
```

```python
from sklearn.dummy import DummyClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

# Création de la baseline
baseline = DummyClassifier(strategy="most_frequent")

# Entraînement
baseline.fit(X_train, y_train)

# Prédictions
y_pred_baseline = baseline.predict(X_test)

# Calcul des métriques
accuracy = accuracy_score(y_test, y_pred_baseline)
precision = precision_score(y_test, y_pred_baseline, zero_division=0)
recall = recall_score(y_test, y_pred_baseline, zero_division=0)
f1 = f1_score(y_test, y_pred_baseline, zero_division=0)

print("Accuracy :", accuracy)
print("Précision :", precision)
print("Rappel :", recall)
print("F1-score :", f1)

print("\nMatrice de confusion :")
print(confusion_matrix(y_test, y_pred_baseline))

print("\nRapport de classification :")
print(classification_report(y_test, y_pred_baseline, zero_division=0))
```

```python
from sklearn.metrics import confusion_matrix

matrice = confusion_matrix(y_test, y_pred_baseline)

print(matrice)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression

modele_logistique = Pipeline([
    ("standardisation", StandardScaler()),
    ("modele", LogisticRegression(
        class_weight="balanced",
        max_iter=300,
        random_state=42
    ))
])

modele_logistique.fit(X_train, y_train)

y_pred_logistique = modele_logistique.predict(X_test)

proba_logistique = modele_logistique.predict_proba(X_test)[:, 1]
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    confusion_matrix,
    classification_report
)

print("Accuracy :", accuracy_score(y_test, y_pred_logistique))

print("Précision :",
      precision_score(y_test, y_pred_logistique, zero_division=0))

print("Rappel :",
      recall_score(y_test, y_pred_logistique, zero_division=0))

print("F1-score :",
      f1_score(y_test, y_pred_logistique, zero_division=0))

print("ROC-AUC :",
      roc_auc_score(y_test, proba_logistique))

print("PR-AUC :",
      average_precision_score(y_test, proba_logistique))

print("\nMatrice de confusion :")
print(confusion_matrix(y_test, y_pred_logistique))

print("\nRapport détaillé :")
print(classification_report(
    y_test,
    y_pred_logistique,
    zero_division=0
))
```

```python
from sklearn.ensemble import RandomForestClassifier

modele_foret = RandomForestClassifier(
    n_estimators=100,
    max_depth=12,
    min_samples_leaf=2,
    class_weight="balanced_subsample",
    n_jobs=-1,
    random_state=42
)

modele_foret.fit(X_train, y_train)

y_pred_foret = modele_foret.predict(X_test)

proba_foret = modele_foret.predict_proba(X_test)[:, 1]
```

```python
print("Accuracy :",
      accuracy_score(y_test, y_pred_foret))

print("Précision :",
      precision_score(y_test, y_pred_foret, zero_division=0))

print("Rappel :",
      recall_score(y_test, y_pred_foret, zero_division=0))

print("F1-score :",
      f1_score(y_test, y_pred_foret, zero_division=0))

print("ROC-AUC :",
      roc_auc_score(y_test, proba_foret))

print("PR-AUC :",
      average_precision_score(y_test, proba_foret))

print("\nMatrice de confusion :")
print(confusion_matrix(y_test, y_pred_foret))
```
